```python
# === CONFIGURA QUI ===
KAGGLE_DATASET_PATH = "/kaggle/input/datasets/lorenzosalis/finetuned-gemma-2b/final" # Ricordati di cambiarlo con la cartellina dei 2 file puliti!
BASE_MODEL_ID       = "google/gemma-2b"

# Tasks. Se vedi che MMLU è troppo lungo anche con il LIMIT, rimuovilo temporaneamente.
TASKS = "arc_challenge,hellaswag,winogrande,truthfulqa_mc2,mmlu"

NUM_FEWSHOT = 0      # 0 = zero-shot; 
LIMIT       = 250    # ATTENZIONE: 250 va benissimo per test rapidi. Su MMLU viene applicato a TUTTE le 57 sottocategorie (quindi 57 * 250 = 14.250 domande). 

# === MAPPE METRICHE (Mancavano nel tuo codice) ===
METRIC_MAP = {
    "arc_challenge": "acc_norm,none", # lm-eval v0.4 usa questo formato
    "hellaswag": "acc_norm,none",
    "winogrande": "acc,none",
    "truthfulqa_mc2": "acc,none",
    "mmlu": "acc,none"
}

OFFICIAL_BASELINE = {
    "arc_challenge": 0.42, 
    "hellaswag": 0.71,
    "winogrande": 0.65,
    "truthfulqa_mc2": 0.44,
    "mmlu": 0.42
}
```

```python
from kaggle_secrets import UserSecretsClient
from huggingface_hub import login

secret = UserSecretsClient()
hf_token = secret.get_secret("HF_TOKEN")
login(token=hf_token)
print("✅ Autenticato su HuggingFace")
```

```python
import torch
import gc
import json
from lm_eval import evaluator
from lm_eval.models.huggingface import HFLM

def free_memory():
    gc.collect()
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

free_memory()
print(f"GPU libera prima del caricamento: {torch.cuda.mem_get_info()[0]/1e9:.2f} GB")

# Inizializzazione modello
model_ft = HFLM(
    pretrained=BASE_MODEL_ID,
    peft=KAGGLE_DATASET_PATH,
    dtype="bfloat16",
    device="cuda",
    # RIMOSSO: max_batch_size=1 (Limitava pesantemente la velocità)
    max_length=1024, # ALZATO: 512 era troppo basso e avrebbe troncato le domande di MMLU
)

print(f"GPU dopo caricamento: {torch.cuda.mem_get_info()[0]/1e9:.2f} GB libera")

# Valutazione
results_ft = evaluator.simple_evaluate(
    model=model_ft,
    tasks=TASKS.split(","),
    num_fewshot=NUM_FEWSHOT,
    batch_size="auto", # MAGIA: Trova il max batch size automaticamente evitando OOM!
    limit=LIMIT,
)

# Salvataggio manuale
with open("/kaggle/working/results_finetuned.json", "w") as f:
    json.dump(results_ft["results"], f, indent=2)

del model_ft
free_memory()
print(f"GPU dopo cleanup: {torch.cuda.mem_get_info()[0]/1e9:.2f} GB libera")
print("Fine-tuned: DONE ✓")

# Stampa dei risultati
print(f"\n{'Task':<20} {'Baseline (Indicativa)':>22} {'Fine-tuned':>12} {'Delta':>8}")
print("-" * 66)

ft_data = results_ft["results"]

for task, metric in METRIC_MAP.items():
    baseline_score = OFFICIAL_BASELINE.get(task)
    
    # lm-eval spesso salva i task compositi come MMLU in modo aggregato o nei subtask
    # Questa try-except serve a catturare il valore esatto
    try:
        # Se presente, cerchiamo il valore esatto della metrica
        ft_score = ft_data[task].get(metric) 
        # Nelle versioni recenti di lm-eval, la metrica potrebbe chiamarsi semplicemente "acc" o "acc_norm"
        if ft_score is None:
            fallback_metric = metric.split(",")[0] # Prende "acc" da "acc,none"
            ft_score = ft_data[task].get(fallback_metric)
    except KeyError:
        ft_score = None

    if baseline_score and ft_score is not None:
        delta = ft_score - baseline_score
        arrow = "▲" if delta > 0 else ("▼" if delta < 0 else "=")
        print(f"{task:<20} {baseline_score:>22.1%} {ft_score:>12.1%} {arrow}{abs(delta):>6.1%}")
    else:
        print(f"{task:<20} {'N/A':>22} {str(ft_score or 'N/A'):>12}")
```
